```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.utils import to_categorical
import numpy as np
import random
import matplotlib.pyplot as plt
import pandas as pd

# Laden und Vorverarbeiten der Daten
data = np.loadtxt('class_128_28x28_CNN.csv', delimiter=',')
labels = data[:, 0].astype(int)
labels = labels.astype(str)
labels = labels.reshape([300, 1])
testlabels = to_categorical(labels)
images = data[:, 1:].reshape(-1, 28, 28, 1) / 255.0

# Aufteilen der Daten in Trainings- und Testdaten
split_index = int(0.8 * len(images))
train_images, test_images = images[:split_index], images[split_index:]
train_labels, test_labels = testlabels[:split_index], testlabels[split_index:]

# Erstellen des CNN-Modells
model = models.Sequential([
    layers.Conv2D(4, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(9, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()

random_image = random.choice(train_images)

# Callback zum Anzeigen der Ergebnisse des Convolution-Kernels pro Epoche
class ConvKernelCallback(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        #random_image = random.choice(train_images)
        conv_layer = model.layers[0]
        kernel_values = conv_layer.get_weights()[0]

        print("Convolution-Kernel Werte:")
        print(kernel_values)

        filter_outputs = tf.keras.Model(inputs=model.inputs, outputs=conv_layer.output)
        filtered_image = filter_outputs(np.expand_dims(random_image, axis=0))

        plt.figure(figsize=(12, 6))
        for i in range(4):
            plt.subplot(1, 5, i + 1)
            plt.imshow(filtered_image[0, :, :, i], cmap='gray')
            plt.axis('off')
        plt.subplot(1, 5, 5)
        plt.imshow(random_image[:, :, 0], cmap='gray')
        plt.axis('off')
        plt.show()

# Modelltraining
history = model.fit(train_images, train_labels, epochs=20, callbacks=[ConvKernelCallback()])

# Vorhersage und Ausgabe
random_index = random.randint(0, len(test_images) - 1)
random_test_image = test_images[random_index]
predicted_label = np.argmax(model.predict(np.expand_dims(random_test_image, axis=0)))

plt.imshow(random_test_image[:, :, 0], cmap='gray')
plt.axis('off')
plt.show()
print(f'Vorhersage: {predicted_label}')

# Berechnung der Modellgenauigkeit
_,accuracy = model.evaluate(test_images, test_labels, verbose=0)
print(f'Modellgenauigkeit: {accuracy:.4f}')
```
