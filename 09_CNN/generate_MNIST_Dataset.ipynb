```python
from codecs import ignore_errors

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Daten einlesen
train = pd.read_csv('Datasets/mnist_train.csv')
print(train.shape)

# neuen leeren dataframe erstellen
df = pd.DataFrame(columns=train.columns)

# für jedes label (0-9)
gesamtmenge = []
for i in range(10):
    # alle Zeilen mit dem label i in einen neuen dataframe kopieren
    teilmenge = train[train.iloc[:, 0] == i].head(800)
    gesamtmenge.append(teilmenge)

df = pd.concat(gesamtmenge, ignore_index=True)
print(df.shape)

# neue spaltennamen
df.columns = ['label'] + [f'pixel_{i}' for i in range(784)]

# neuen datensatz speichern:
df.to_csv('Datasets/mnist_train_800.csv', index=False)
```
